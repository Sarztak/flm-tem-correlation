# Object masks in images from prompts with SAM 2

Segment Anything Model 2 (SAM 2) predicts object masks given prompts that indicate the desired object. The model first converts the image into an image embedding that allows high quality masks to be efficiently produced from a prompt.

The `SAM2ImagePredictor` class provides an easy interface to the model for prompting the model. It allows the user to first set an image using the `set_image` method, which calculates the necessary image embeddings. Then, prompts can be provided via the `predict` method to efficiently predict masks from those prompts. The model can take as input both point and box prompts, as well as masks from the previous iteration of prediction.

Necessary imports and helper functions for displaying points, boxes, and masks.

```python
import os
# if using Apple MPS, fall back to CPU for unsupported ops
os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "1"
import numpy as np
import torch
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# select the device for computation
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")
print(f"using device: {device}")

if device.type == "cuda":
    # use bfloat16 for the entire notebook
    torch.autocast("cuda", dtype=torch.bfloat16).__enter__()
    # turn on tfloat32 for Ampere GPUs (https://pytorch.org/docs/stable/notes/cuda.html#tensorfloat-32-tf32-on-ampere-devices)
    if torch.cuda.get_device_properties(0).major >= 8:
        torch.backends.cuda.matmul.allow_tf32 = True
        torch.backends.cudnn.allow_tf32 = True
elif device.type == "mps":
    print(
        "\nSupport for MPS devices is preliminary. SAM 2 is trained with CUDA and might "
        "give numerically different outputs and sometimes degraded performance on MPS. "
        "See e.g. https://github.com/pytorch/pytorch/issues/84936 for a discussion."
    )
```

```python
np.random.seed(3)

def show_mask(mask, ax, random_color=False, borders = True):
    if random_color:
        color = np.concatenate([np.random.random(3), np.array([0.6])], axis=0)
    else:
        color = np.array([30/255, 144/255, 255/255, 0.6])
    h, w = mask.shape[-2:]
    mask = mask.astype(np.uint8)
    mask_image =  mask.reshape(h, w, 1) * color.reshape(1, 1, -1)
    if borders:
        import cv2
        contours, _ = cv2.findContours(mask,cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
        # Try to smooth contours
        contours = [cv2.approxPolyDP(contour, epsilon=0.01, closed=True) for contour in contours]
        mask_image = cv2.drawContours(mask_image, contours, -1, (1, 1, 1, 0.5), thickness=2)
    ax.imshow(mask_image)

def show_points(coords, labels, ax, marker_size=375):
    pos_points = coords[labels==1]
    neg_points = coords[labels==0]
    ax.scatter(pos_points[:, 0], pos_points[:, 1], color='green', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)
    ax.scatter(neg_points[:, 0], neg_points[:, 1], color='red', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)

def show_box(box, ax):
    x0, y0 = box[0], box[1]
    w, h = box[2] - box[0], box[3] - box[1]
    ax.add_patch(plt.Rectangle((x0, y0), w, h, edgecolor='green', facecolor=(0, 0, 0, 0), lw=2))

def show_masks(image, masks, scores, point_coords=None, box_coords=None, input_labels=None, borders=True, figsize=(5, 5)):
    for i, (mask, score) in enumerate(zip(masks, scores)):
        plt.figure(figsize=figsize)
        plt.imshow(image)
        show_mask(mask, plt.gca(), borders=borders)
        if point_coords is not None:
            assert input_labels is not None
            show_points(point_coords, input_labels, plt.gca())
        if box_coords is not None:
            # boxes
            show_box(box_coords, plt.gca())
        if len(scores) > 1:
            plt.title(f"Mask {i+1}, Score: {score:.3f}", fontsize=18)
        plt.axis('off')
        plt.show()
```

## Selecting objects with SAM 2

First, load the SAM 2 model and predictor. Change the path below to point to the SAM 2 checkpoint. Running on CUDA and using the default model are recommended for best results.

```python
from sam2.build_sam import build_sam2
from sam2.sam2_image_predictor import SAM2ImagePredictor

sam2_checkpoint = "../checkpoints/sam2.1_hiera_large.pt"
model_cfg = "configs/sam2.1/sam2.1_hiera_l.yaml"

sam2_model = build_sam2(model_cfg, sam2_checkpoint, device=device)

predictor = SAM2ImagePredictor(sam2_model)
```

Process the image to produce an image embedding by calling `SAM2ImagePredictor.set_image`. `SAM2ImagePredictor` remembers this embedding and will use it for subsequent mask prediction.

```python
predictor.set_image(image)
```

To select the truck, choose a point on it. Points are input to the model in (x,y) format and come with labels 1 (foreground point) or 0 (background point). Multiple points can be input; here we use only one. The chosen point will be shown as a star on the image.

```python
input_point = np.array([[550, 800]])
input_label = np.array([1])
```

```python
plt.figure(figsize=(4, 4))
plt.imshow(image)
show_points(input_point, input_label, plt.gca())
plt.axis('on')
plt.show()
```

```python
print(predictor._features["image_embed"].shape, predictor._features["image_embed"][-1].shape)
```

Predict with `SAM2ImagePredictor.predict`. The model returns masks, quality predictions for those masks, and low resolution mask logits that can be passed to the next iteration of prediction.

```python
masks, scores, logits = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    multimask_output=True,
)
sorted_ind = np.argsort(scores)[::-1]
masks = masks[sorted_ind]
scores = scores[sorted_ind]
logits = logits[sorted_ind]
```

With `multimask_output=True` (the default setting), SAM 2 outputs 3 masks, where `scores` gives the model's own estimation of the quality of these masks. This setting is intended for ambiguous input prompts, and helps the model disambiguate different objects consistent with the prompt. When `False`, it will return a single mask. For ambiguous prompts such as a single point, it is recommended to use `multimask_output=True` even if only a single mask is desired; the best single mask can be chosen by picking the one with the highest score returned in `scores`. This will often result in a better mask.

```python
masks.shape  # (number_of_masks) x H x W
```

```python
show_masks(image, masks, scores, point_coords=input_point, input_labels=input_label, borders=True)
```

## Specifying a specific object with additional points

The single input point is ambiguous, and the model has returned multiple objects consistent with it. To obtain a single object, multiple points can be provided. If available, a mask from a previous iteration can also be supplied to the model to aid in prediction. When specifying a single object with multiple prompts, a single mask can be requested by setting `multimask_output=False`.

```python
input_point = np.array([[550, 800], [550, 1000]])
input_label = np.array([1, 1])

mask_input = logits[np.argmax(scores), :, :]  # Choose the model's best mask
```

```python
masks, scores, _ = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    mask_input=mask_input[None, :, :],
    multimask_output=False,
)
```

```python
masks.shape
```

```python
show_masks(image, masks, scores, point_coords=input_point, input_labels=input_label)
```

```python
input_point = np.array([[800, 1230], [1000, 1000]])
input_label = np.array([1, 0])

mask_input = logits[np.argmax(scores), :, :]  # Choose the model's best mask
```

```python
masks, scores, _ = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    mask_input=mask_input[None, :, :],
    multimask_output=False,
)
```

```python
show_masks(image, masks, scores, point_coords=input_point, input_labels=input_label)
```

## End-to-end batched inference
If all prompts are available in advance, it is possible to run SAM 2 directly in an end-to-end fashion. This also allows batching over images.

```python
image1 = image  # truck.jpg from above
image1_boxes = np.array([
    [75, 275, 1725, 850],
    [425, 600, 700, 875],
    [1375, 550, 1650, 800],
    [1240, 675, 1400, 750],
])

image2 = Image.open('images/groceries.jpg')
image2 = np.array(image2.convert("RGB"))
image2_boxes = np.array([
    [450, 170, 520, 350],
    [350, 190, 450, 350],
    [500, 170, 580, 350],
    [580, 170, 640, 350],
])

img_batch = [image1, image2]
boxes_batch = [image1_boxes, image2_boxes]
```

```python
predictor.set_image_batch(img_batch)
```

```python
masks_batch, scores_batch, _ = predictor.predict_batch(
    None,
    None,
    box_batch=boxes_batch,
    multimask_output=False
)
```

```python
for image, boxes, masks in zip(img_batch, boxes_batch, masks_batch):
    plt.figure(figsize=(10, 10))
    plt.imshow(image)
    for mask in masks:
        show_mask(mask.squeeze(0), plt.gca(), random_color=True)
    for box in boxes:
        show_box(box, plt.gca())
```

Similarly, we can have a batch of point prompts defined over a batch of images

```python
image1 = image  # truck.jpg from above
image1_pts = np.array([
    [[500, 375]],
    [[650, 750]]
    ]) # Bx1x2 where B corresponds to number of objects
image1_labels = np.array([[1], [1]])

image2_pts = np.array([
    [[400, 300]],
    [[630, 300]],
])
image2_labels = np.array([[1], [1]])

pts_batch = [image1_pts, image2_pts]
labels_batch = [image1_labels, image2_labels]
```

```python
masks_batch, scores_batch, _ = predictor.predict_batch(pts_batch, labels_batch, box_batch=None, multimask_output=True)

# Select the best single mask per object
best_masks = []
for masks, scores in zip(masks_batch,scores_batch):
    best_masks.append(masks[range(len(masks)), np.argmax(scores, axis=-1)])
```

```python
for image, points, labels, masks in zip(img_batch, pts_batch, labels_batch, best_masks):
    plt.figure(figsize=(10, 10))
    plt.imshow(image)
    for mask in masks:
        show_mask(mask, plt.gca(), random_color=True)
    show_points(points, labels, plt.gca())
```

```python
%%writefile sam2_app.py
import streamlit as st
import numpy as np
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt
import torch
import io
import logging
logging.getLogger("streamlit").setLevel(logging.ERROR)

from sam2.build_sam import build_sam2
from sam2.sam2_image_predictor import SAM2ImagePredictor
from streamlit_image_coordinates import streamlit_image_coordinates

st.set_page_config(layout="wide")
st.title("SAM2 Interactive Segmentation")

SAM2_CHECKPOINT = "../checkpoints/sam2.1_hiera_large.pt"
MODEL_CFG       = "configs/sam2.1/sam2.1_hiera_l.yaml"
DEVICE          = "cuda" if torch.cuda.is_available() else "cpu"
DISPLAY_WIDTH   = 640
DISPLAY_HEIGHT  = 480

@st.cache_resource
def load_predictor():
    sam2_model = build_sam2(MODEL_CFG, SAM2_CHECKPOINT, device=DEVICE)
    return SAM2ImagePredictor(sam2_model)

predictor = load_predictor()

for key, val in {
    "points": [], "labels": [], "logits": None,
    "last_click": None, "result_img": None,
    "last_score": None, "last_mask": None, "phase": "collect"
}.items():
    st.session_state.setdefault(key, val)

uploaded = st.file_uploader("Upload an image", type=["jpg", "jpeg", "png"])

if uploaded:
    image_pil      = Image.open(uploaded).convert("RGB")
    image_np       = np.array(image_pil)
    orig_w, orig_h = image_pil.size
    scale_x        = orig_w / DISPLAY_WIDTH
    scale_y        = orig_h / DISPLAY_HEIGHT
    display_img    = image_pil.resize((DISPLAY_WIDTH, DISPLAY_HEIGHT))

    def draw_points_on_image(base_img):
        img  = base_img.copy()
        draw = ImageDraw.Draw(img)
        for (dx, dy), lbl in zip(st.session_state.points, st.session_state.labels):
            color  = (0, 220, 0) if lbl == 1 else (220, 0, 0)
            radius = 8
            draw.ellipse(
                [(dx - radius, dy - radius), (dx + radius, dy + radius)],
                fill=color, outline="white", width=2
            )
        return img

    def run_sam2():
        predictor.set_image(image_np)
        orig_coords = np.array([
            [int(x * scale_x), int(y * scale_y)]
            for x, y in st.session_state.points
        ])
        labs = np.array(st.session_state.labels)

        predict_kwargs = dict(point_coords=orig_coords, point_labels=labs, multimask_output=True)
        if st.session_state.logits is not None:
            predict_kwargs.update(
                mask_input=st.session_state.logits[None, :, :],
                multimask_output=False,
            )

        with torch.inference_mode():
            masks, scores, logits = predictor.predict(**predict_kwargs)

        best_idx = int(np.argmax(scores))
        st.session_state.logits = logits[best_idx]
        best_mask = masks[best_idx]

        fig, ax = plt.subplots(figsize=(DISPLAY_WIDTH / 100, DISPLAY_HEIGHT / 100), dpi=100)
        ax.imshow(image_np)
        ax.imshow(best_mask, alpha=0.4, cmap="cool")
        ax.contour(best_mask, colors="cyan", linewidths=1.5)
        pos = orig_coords[labs == 1]
        neg = orig_coords[labs == 0]
        if len(pos): ax.scatter(pos[:, 0], pos[:, 1], c="lime", marker="*", s=350, edgecolors="white", linewidths=1.2, zorder=5)
        if len(neg): ax.scatter(neg[:, 0], neg[:, 1], c="red",  marker="*", s=350, edgecolors="white", linewidths=1.2, zorder=5)
        ax.axis("off")
        plt.tight_layout(pad=0)
        buf = io.BytesIO()
        plt.savefig(buf, format="png", dpi=100, bbox_inches="tight", pad_inches=0)
        plt.close()
        buf.seek(0)
        return buf.read(), scores[best_idx], best_mask

    left, right = st.columns([1, 1], gap="large")

    with left:
        st.subheader("Controls")

        mode      = st.radio("Click mode", ["Foreground ✅", "Background ❌", "Delete 🗑️"], horizontal=True)
        label_val = 1 if "Foreground" in mode else 0

        def handle_click(click):
            if not click or click == st.session_state.last_click:
                return False
            st.session_state.last_click = click
            cx, cy = click["x"], click["y"]
            if "Delete" in mode:
                if st.session_state.points:
                    dists   = [((cx - x) ** 2 + (cy - y) ** 2) for x, y in st.session_state.points]
                    closest = int(np.argmin(dists))
                    st.session_state.points.pop(closest)
                    st.session_state.labels.pop(closest)
                    st.session_state.logits = None
            else:
                st.session_state.points.append([cx, cy])
                st.session_state.labels.append(label_val)
            return True

        if st.session_state.phase == "collect":
            st.write("👆 Click on the image to place points")
            annotated = draw_points_on_image(display_img)
            click = streamlit_image_coordinates(annotated, key="img_click", width=DISPLAY_WIDTH)
            if handle_click(click):
                st.rerun()

            st.caption(f"Points: {len(st.session_state.points)}")
            st.divider()

            c1, c2 = st.columns(2)
            with c1:
                if st.button("↩️ Undo") and st.session_state.points:
                    st.session_state.points.pop()
                    st.session_state.labels.pop()
                    st.session_state.logits = None
                    st.rerun()
            with c2:
                if st.button("🗑️ Clear All"):
                    st.session_state.points     = []
                    st.session_state.labels     = []
                    st.session_state.logits     = None
                    st.session_state.last_click = None
                    st.session_state.result_img = None
                    st.rerun()

            st.divider()
            if st.button("▶️ Run SAM2", disabled=len(st.session_state.points) == 0, type="primary"):
                st.session_state.phase = "result"
                st.rerun()

        else:
            st.write("✅ Add or delete points to refine, then press Refine.")
            annotated = draw_points_on_image(display_img)
            click = streamlit_image_coordinates(annotated, key="img_click_refine", width=DISPLAY_WIDTH)
            if handle_click(click):
                st.rerun()

            st.caption(f"Points: {len(st.session_state.points)}")
            st.divider()

            c1, c2 = st.columns(2)
            with c1:
                if st.button("↩️ Undo") and st.session_state.points:
                    st.session_state.points.pop()
                    st.session_state.labels.pop()
                    st.session_state.logits = None
                    st.rerun()
            with c2:
                if st.button("🔄 Restart"):
                    for k in ["points", "labels", "logits", "last_click", "result_img", "last_mask", "last_score"]:
                        st.session_state[k] = [] if k in ["points", "labels"] else None
                    st.session_state.phase = "collect"
                    st.rerun()

            st.divider()
            if st.button("▶️ Refine Mask", type="primary", disabled=len(st.session_state.points) == 0):
                result_bytes, score, mask = run_sam2()
                st.session_state.result_img = result_bytes
                st.session_state.last_score = score
                st.session_state.last_mask  = mask
                st.rerun()

            if st.session_state.last_mask is not None:
                mask_img = Image.fromarray((st.session_state.last_mask * 255).astype(np.uint8))
                buf = io.BytesIO()
                mask_img.save(buf, format="PNG")
                st.download_button("💾 Save Mask", buf.getvalue(), "mask.png", "image/png")

    with right:
        st.subheader("Output")
        if st.session_state.phase == "collect":
            st.image(display_img, width=DISPLAY_WIDTH, caption="Output will appear here after running SAM2")
        else:
            if st.session_state.result_img is None:
                result_bytes, score, mask = run_sam2()
                st.session_state.result_img = result_bytes
                st.session_state.last_score = score
                st.session_state.last_mask  = mask
            st.image(st.session_state.result_img, width=DISPLAY_WIDTH)
            st.caption(f"Mask score: {st.session_state.last_score:.3f} | Points: {len(st.session_state.points)}")
```
